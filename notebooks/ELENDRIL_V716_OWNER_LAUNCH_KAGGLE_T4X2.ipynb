```python
"""Owner-authorized V16 launch: diagnose attached Kaggle inputs before GPU training.

All input files are read-only. The diagnostic shows exactly which datasets
and ZIP archives are present and admits only the historical FULL ZIP SHA-256.
A NO-GO is a custody failure, not a neural-network or GPU learning result.
"""
from hashlib import sha256
from pathlib import Path
import json
import subprocess

PINNED_PARENT_SHA256 = "1826c85b5a858d2a600661f7e5916f0fa812c55a8d53c37013c695df28a15d19"
PARENT_CORE_RUN_ID = "20c80f66e76e4b589aed2053afbddffc"
PARENT_CORE_SOURCE_SHA = "6104ffd7134cb4cf4b3f340bd060d24a1e2dd200"
PARENT_CORE_BYTES = 1958029910
INPUT_DIAGNOSTIC = Path("/kaggle/working/V16_INPUT_DIAGNOSTIC.json")


def _file_sha256(path):
    digest = sha256()
    with Path(path).open("rb") as handle:
        while True:
            block = handle.read(8 * 1024 * 1024)
            if not block:
                break
            digest.update(block)
    return digest.hexdigest()


def _find_parent_archive(root, expected_sha256, diagnostic_path=None):
    """Inventory and hash Kaggle ZIPs, save a report, refuse unverified inputs."""
    root = Path(root)
    root_exists = root.is_dir()
    report = {
        "schema": "v16-kaggle-input-diagnostic-v1",
        "expected_parent_sha256": expected_sha256,
        "expected_parent_run_id": PARENT_CORE_RUN_ID,
        "expected_parent_source": PARENT_CORE_SOURCE_SHA,
        "expected_parent_bytes": PARENT_CORE_BYTES,
        "input_root": str(root),
        "attached_datasets": [],
        "file_count": 0,
        "file_examples": [],
        "zip_archives": [],
        "status": "INSPECTING",
    }

    def publish():
        if diagnostic_path is not None:
            output = Path(diagnostic_path)
            output.parent.mkdir(parents=True, exist_ok=True)
            output.write_text(json.dumps(report, indent=2, sort_keys=True) + "\n",
                              encoding="utf-8")
            print("V16 input diagnostic saved:", output, flush=True)

    if not root_exists:
        report["status"] = "MISSING_INPUT_ROOT"
        publish()
        raise RuntimeError("V16 INPUT NO-GO: /kaggle/input directory is missing. "
                           "Attach the genuine V7.1 CORE FULL ZIP via Kaggle Add Input.")

    report["attached_datasets"] = sorted(
        entry.name for entry in root.iterdir() if entry.is_dir() and not entry.is_symlink())
    candidates = []
    for path in root.rglob("*"):
        if not path.is_file() or path.is_symlink():
            continue
        report["file_count"] += 1
        if len(report["file_examples"]) < 25:
            report["file_examples"].append(str(path.relative_to(root)))
        if path.suffix.lower() == ".zip":
            candidates.append(path)
    candidates.sort(key=lambda path: str(path))
    print("V16 REQUIRED original V7.1 CORE FULL ZIP:", PARENT_CORE_RUN_ID,
          "| expected bytes:", PARENT_CORE_BYTES, "| SHA-256:", expected_sha256,
          flush=True)
    print("V16 attached Kaggle datasets:", report["attached_datasets"], flush=True)
    print("V16 input files:", report["file_count"],
          "| ZIP candidates:", len(candidates), flush=True)
    for name in report["file_examples"][:15]:
        print("V16 input example:", name, flush=True)

    if not candidates:
        report["status"] = "NO_ZIP_ARCHIVES"
        publish()
        raise RuntimeError(
            "V16 INPUT NO-GO: no ZIP files found under /kaggle/input. "
            "Attached extracted V7.15 scale-S8 files or a Kaggle notebook output "
            "are NOT proof of the required historical V7.1 CORE FULL ZIP. "
            "Use Add Input to attach the original V7.1 CORE FULL ZIP, "
            "run 20c80f66e76e4b589aed2053afbddffc "
            "(1,958,029,910 bytes). "
            "See V16_INPUT_DIAGNOSTIC.json for the actual input inventory.")

    matching = None
    for path in candidates:
        digest = _file_sha256(path)
        same = digest == expected_sha256
        record = {
            "path": str(path),
            "sha256": digest,
            "byte_size": path.stat().st_size,
            "parent_full_match": same,
            "expected_core_byte_size_match": path.stat().st_size == PARENT_CORE_BYTES,
            "looks_like_wrong_s8_stage": "scale-s8" in str(path).lower(),
        }
        report["zip_archives"].append(record)
        print("V16 ZIP:", path, "| bytes:", record["byte_size"],
              "| SHA-256:", digest, "| MATCH:", same, flush=True)
        if same:
            matching = path
    if matching is None:
        report["status"] = "ZIP_PRESENT_BUT_HASH_MISMATCH"
        report["wrong_v715_scale_s8_detected"] = any(
            row["looks_like_wrong_s8_stage"] for row in report["zip_archives"])
        if report["wrong_v715_scale_s8_detected"]:
            print("V16 WRONG INPUT: attached V7.15 scale-S8 outputs, "
                  "not the V7.1 CORE FULL archive.", flush=True)
        publish()
        raise RuntimeError(
            "V16 INPUT NO-GO: none of the " + str(len(candidates)) +
            " attached ZIP files matches the historical V7.1 CORE FULL "
            "SHA-256 " + expected_sha256 + ". An ANALYSIS ZIP, an extracted "
            "dataset, or another V7.15 stage is not a substitute. "
            "Find the earlier V7.1 CORE Kaggle output, not the V7.15 scale-S8 "
            "notebook output. Required ELENDRIL_V71_FULL.zip is exactly "
            "1,958,029,910 bytes. See V16_INPUT_DIAGNOSTIC.json for observed "
            "SHA-256 values.")
    report["status"] = "VERIFIED_PARENT_FULL"
    report["selected_parent_archive"] = str(matching)
    publish()
    return matching


def _auto_fetch_original_core(expected_sha256):
    """Try Kaggle's native notebook-output API; no owner Add Input required.

    The original V7.1 CORE notebook was observed at this exact Kaggle handle.
    Download one specific historical FULL ZIP, not latest unrelated output.
    Fail closed if it is unavailable, private, missing or hash-mismatched.
    """
    handle = "ankitrajm7c/elendril-v71-kaggle-t4x2"
    relative = "elendril-v71-6104ffd7134c-core/exports/ELENDRIL_V71_FULL.zip"
    try:
        import kagglehub
    except ImportError as exc:
        raise RuntimeError(
            "V16 AUTO-FETCH NO-GO: kagglehub is unavailable in this Kaggle runtime. "
            "No dataset has been substituted.") from exc
    print("V16 attempting automatic read-only attachment from Kaggle notebook:",
          handle, "path:", relative, flush=True)
    try:
        downloaded = kagglehub.notebook_output_download(handle, path=relative)
    except Exception as exc:
        raise RuntimeError(
            "V16 AUTO-FETCH NO-GO: Kaggle refused runtime attachment of the "
            "recorded V7.1 CORE notebook output. Kaggle cannot attach notebook outputs inside "
            "non-interactive Save & Run All sessions. Remedy: open this notebook in the interactive "
            "editor, use Add Input > Notebook Output to attach the output of ankitrajm7c/"
            "elendril-v71-kaggle-t4x2 (V7.1 CORE run 20c80f66e76e4b589aed2053afbddffc), then "
            "Save & Run All. Original data has NOT been replaced.") from exc
    target = Path(downloaded)
    if target.is_dir():
        candidates = sorted(target.rglob("ELENDRIL_V71_FULL.zip"))
        if len(candidates) != 1:
            raise RuntimeError("V16 AUTO-FETCH NO-GO: downloaded parent file is ambiguous")
        target = candidates[0]
    if not target.is_file() or target.is_symlink():
        raise RuntimeError("V16 AUTO-FETCH NO-GO: Kaggle returned no regular parent FULL ZIP")
    if target.stat().st_size != PARENT_CORE_BYTES:
        raise RuntimeError(
            "V16 AUTO-FETCH NO-GO: remote FULL archive size mismatched the "
            "recorded V7.1 CORE parent; got " + str(target.stat().st_size))
    actual = _file_sha256(target)
    if actual != expected_sha256:
        raise RuntimeError("V16 AUTO-FETCH NO-GO: remote parent SHA-256 mismatch: " + actual)
    print("V16 AUTO-FETCH VERIFIED V7.1 CORE FULL:", target, flush=True)
    return target


# EXECUTE OWNER PREFLIGHT (no GPU training begins before this passes)
if RUN_STUDY is not True:
    raise RuntimeError("OWNER LAUNCH CONFIG ERROR: the owner launch notebook requires RUN_STUDY=True")
if PHASE != "mechanism" or ARM_SUBSET != "H2B-NORM" or SEED_SUBSET != 7:
    raise RuntimeError("OWNER LAUNCH CONFIG ERROR: expected H2B-NORM seed-7 screening")
if PARENT_OUTPUT is not None:
    raise RuntimeError("OWNER LAUNCH CONFIG ERROR: this launch profile requires the recorded parent FULL ZIP")
if PARENT_ARCHIVE is None:
    try:
        PARENT_ARCHIVE = _find_parent_archive(
            "/kaggle/input", PINNED_PARENT_SHA256, diagnostic_path=INPUT_DIAGNOSTIC)
    except RuntimeError as input_error:
        print("V16 original parent not mounted; trying automatic KaggleHub attachment.",
              flush=True)
        try:
            PARENT_ARCHIVE = _auto_fetch_original_core(PINNED_PARENT_SHA256)
        except RuntimeError as fetch_error:
            raise RuntimeError(
                "V16 PARENT UNAVAILABLE: neither existing read-only Kaggle inputs "
                "nor the automatic original V7.1 CORE notebook-output fetch "
                "provided the verified FULL archive. Attached-input diagnosis: " +
                str(input_error) + ". Automatic-fetch diagnosis: " +
                str(fetch_error)) from fetch_error
        if INPUT_DIAGNOSTIC.is_file():
            report = json.loads(INPUT_DIAGNOSTIC.read_text(encoding="utf-8"))
            report.update(status="AUTO_FETCHED_PARENT_VERIFIED",
                          selected_parent_archive=str(PARENT_ARCHIVE),
                          auto_fetch_source="ankitrajm7c/elendril-v71-kaggle-t4x2")
            INPUT_DIAGNOSTIC.write_text(
                json.dumps(report, indent=2, sort_keys=True) + "\n",
                encoding="utf-8")
else:
    PARENT_ARCHIVE = Path(PARENT_ARCHIVE)
    if not PARENT_ARCHIVE.is_file() or PARENT_ARCHIVE.is_symlink():
        raise RuntimeError("V16 INPUT NO-GO: configured PARENT_ARCHIVE does not exist as a normal file")
    if _file_sha256(PARENT_ARCHIVE) != PINNED_PARENT_SHA256:
        raise RuntimeError("V16 INPUT NO-GO: configured parent ZIP SHA-256 does not match historical record")
print("V16 VERIFIED parent FULL ZIP:", PARENT_ARCHIVE, flush=True)

try:
    gpu = subprocess.run(
        ["nvidia-smi", "--query-gpu=name,memory.free",
         "--format=csv,noheader,nounits"],
        capture_output=True, text=True, timeout=30, check=True)
except (OSError, subprocess.SubprocessError) as exc:
    raise RuntimeError("V16 HARDWARE NO-GO: choose GPU T4 x2 in Kaggle settings.") from exc
gpu_lines = [line.strip() for line in gpu.stdout.splitlines() if line.strip()]
if len(gpu_lines) != 2 or any("T4" not in line.upper() for line in gpu_lines):
    raise RuntimeError("V16 HARDWARE NO-GO: require two visible NVIDIA T4 GPUs; "
                       "observed: " + repr(gpu_lines))
print("V16 INPUT + VISIBLE GPU PRECHECK PASSED", flush=True)
print("Starting registered H2B-NORM / seed 7 screening with pinned V16 code", flush=True)
```
