```python
"""Owner-authorized Kaggle launch: explicit environment and parent-custody checks.

Runs before GitHub checkout or any optimizer update. All candidate ZIPs are
read-only; only an exact SHA-256 match is allowed to become PARENT_ARCHIVE.
"""
from hashlib import sha256
from pathlib import Path
import subprocess

PINNED_PARENT_SHA256 = "1826c85b5a858d2a600661f7e5916f0fa812c55a8d53c37013c695df28a15d19"


def _file_sha256(path):
    digest = sha256()
    with Path(path).open("rb") as handle:
        while True:
            block = handle.read(8 * 1024 * 1024)
            if not block:
                break
            digest.update(block)
    return digest.hexdigest()


def _find_parent_archive(root, expected_sha256):
    """Search uploaded Kaggle inputs; reject aliases and mismatched archives."""
    root = Path(root)
    if not root.is_dir():
        raise RuntimeError(
            "V16 INPUT NO-GO: /kaggle/input is missing. Attach the authentic "
            "V7.15 FULL result ZIP as Kaggle input before running.")
    candidates = sorted(
        (path for path in root.rglob("*")
         if path.is_file() and not path.is_symlink()
         and path.suffix.lower() == ".zip"),
        key=lambda path: str(path))
    if not candidates:
        raise RuntimeError(
            "V16 INPUT NO-GO: no ZIP files were found under /kaggle/input. "
            "Attach the genuine V7.15 FULL ZIP to the notebook via Add Input.")
    for path in candidates:
        print("V16 checking read-only input ZIP:", path, flush=True)
        if _file_sha256(path) == expected_sha256:
            return path
    raise RuntimeError(
        "V16 INPUT NO-GO: none of the " + str(len(candidates)) +
        " attached ZIP files matches the required V7.15 parent FULL "
        "SHA-256 " + expected_sha256 + ". An ANALYSIS ZIP is not a substitute. "
        "Attach the original FULL archive; do not bypass scientific custody.")


# EXECUTE OWNER PREFLIGHT (no GPU training begins in this cell)
if RUN_STUDY is not True:
    raise RuntimeError("OWNER LAUNCH CONFIG ERROR: this separate launch notebook must have RUN_STUDY=True")
if PHASE != "mechanism" or ARM_SUBSET != "H2B-NORM" or SEED_SUBSET != 7:
    raise RuntimeError("OWNER LAUNCH CONFIG ERROR: expected the registered first screening slice H2B-NORM / seed 7")
if PARENT_OUTPUT is not None:
    raise RuntimeError("OWNER LAUNCH CONFIG ERROR: attach the recorded parent FULL archive; this launch profile does not use extracted PARENT_OUTPUT")
if PARENT_ARCHIVE is None:
    PARENT_ARCHIVE = _find_parent_archive("/kaggle/input", PINNED_PARENT_SHA256)
else:
    PARENT_ARCHIVE = Path(PARENT_ARCHIVE)
    if not PARENT_ARCHIVE.is_file() or PARENT_ARCHIVE.is_symlink():
        raise RuntimeError("V16 INPUT NO-GO: configured PARENT_ARCHIVE does not exist as a normal file")
    if _file_sha256(PARENT_ARCHIVE) != PINNED_PARENT_SHA256:
        raise RuntimeError("V16 INPUT NO-GO: configured parent ZIP SHA-256 does not match historical record")
print("V16 parent FULL ZIP SHA-256 verified:", PARENT_ARCHIVE, flush=True)

try:
    gpu = subprocess.run(
        ["nvidia-smi", "--query-gpu=name,memory.free",
         "--format=csv,noheader,nounits"],
        capture_output=True, text=True, timeout=30, check=True)
except (OSError, subprocess.SubprocessError) as exc:
    raise RuntimeError(
        "V16 HARDWARE NO-GO: Kaggle GPU runtime not available. "
        "Choose accelerator GPU T4 x2 in notebook settings.") from exc
gpu_lines = [line.strip() for line in gpu.stdout.splitlines() if line.strip()]
if len(gpu_lines) != 2 or any("T4" not in line.upper() for line in gpu_lines):
    raise RuntimeError(
        "V16 HARDWARE NO-GO: expected two visible NVIDIA T4 GPUs. "
        "Actual nvidia-smi rows: " + repr(gpu_lines))
print("V16 INPUT + VISIBLE GPU PRECHECK PASSED", flush=True)
print("Starting registered H2B-NORM / seed 7 mechanism screening via pinned V16 code", flush=True)
```
