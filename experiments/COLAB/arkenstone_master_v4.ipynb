Run this notebook in **Colab with a T4 GPU**. It intentionally keeps experiment logic in a syntax-checked repository script so notebook indentation/caching cannot corrupt the campaign.

```python
import os, subprocess, sys, pathlib, torch
assert torch.cuda.is_available(), 'No CUDA GPU. Runtime -> Change runtime type -> T4 GPU.'
print('GPU:', torch.cuda.get_device_name(0), '| torch', torch.__version__)
repo = pathlib.Path('/content/An-Ra-the-new-AGI')
if not repo.exists():
    subprocess.run(['git','clone','--depth','1','--branch','Arkenstone','https://github.com/dhurv0045com-spec/An-Ra-the-new-AGI.git',str(repo)], check=True)
else:
    subprocess.run(['git','-C',str(repo),'fetch','origin','Arkenstone'], check=True)
    subprocess.run(['git','-C',str(repo),'checkout','Arkenstone'], check=True)
    subprocess.run(['git','-C',str(repo),'reset','--hard','origin/Arkenstone'], check=True)
runner = repo / 'experiments/COLAB/run_master_v4.py'
subprocess.run([sys.executable,'-m','py_compile',str(runner)], check=True)
print('Syntax check PASS:', runner)
print('Branch HEAD:', subprocess.check_output(['git','-C',str(repo),'rev-parse','HEAD'], text=True).strip())
```
